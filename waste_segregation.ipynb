```python
# -*- coding: utf-8 -*-
"""waste_segregation_classifier.ipynb

Automatically generated by Colab.

Waste Segregation AI — Eco-Sorter
Classifies trash into: Cardboard, Glass, Metal, Paper, Plastic, Trash
"""

# ╔══════════════════════════════════════════════════════════════════╗
# ║   FULL PIPELINE — Single Cell (GPU)                             ║
# ║   Installs → Downloads Dataset → Trains → Saves → Downloads    ║
# ╚══════════════════════════════════════════════════════════════════╝

# ── 1. Install ────────────────────────────────────────────────────
import subprocess
subprocess.run(['pip', 'install', 'kaggle', 'tensorflow',
                'scikit-learn', 'tqdm', 'opencv-python-headless',
                'matplotlib', 'Pillow', '-q'], check=True)

# ── 2. Imports ────────────────────────────────────────────────────
import os, json, zipfile, warnings, io, base64, glob
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import tensorflow as tf
import cv2
from PIL import Image
from tqdm import tqdm
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
from tensorflow.keras import layers, models, optimizers, callbacks, regularizers
from tensorflow.keras.applications import EfficientNetB3
import ipywidgets as widgets
from ipywidgets import VBox, HBox, Button, Output, HTML, FileUpload, Layout
import IPython.display as ipd

warnings.filterwarnings('ignore')
print(f'✅ TF {tf.__version__} | GPU: {tf.config.list_physical_devices("GPU")}')

# ── 3. Kaggle API setup ───────────────────────────────────────────
KAGGLE_USERNAME = "siddhikapujari"   # ← PASTE YOUR USERNAME
KAGGLE_KEY      = "KGAT_49e2768719aef085dcdcbf986aba9ec8"    # ← PASTE YOUR API KEY

os.makedirs('/root/.kaggle', exist_ok=True)
with open('/root/.kaggle/kaggle.json', 'w') as f:
    json.dump({"username": KAGGLE_USERNAME, "key": KAGGLE_KEY}, f)
os.chmod('/root/.kaggle/kaggle.json', 0o600)
print('✅ Kaggle credentials set')

# ── 4. Download & unzip dataset ───────────────────────────────────
#    Garbage Classification Dataset — 2,527 images, 6 categories
#    Cardboard(403) Glass(501) Metal(410) Paper(594) Plastic(482) Trash(137)
os.makedirs('/content/waste_data', exist_ok=True)
os.system('kaggle datasets download -d asdasdasasdas/garbage-classification '
          '-p /content/waste_data -q')

zip_files = glob.glob('/content/waste_data/*.zip')
for zf in zip_files:
    print(f'📦 Extracting {os.path.basename(zf)}...')
    with zipfile.ZipFile(zf, 'r') as z:
        z.extractall('/content/waste_data')
print('✅ Dataset downloaded and extracted')

# ── 5. Auto-detect class folders ─────────────────────────────────
WASTE_CLASSES = ['cardboard', 'glass', 'metal', 'paper', 'plastic', 'trash']
DATA_ROOT     = None

for root, dirs, files in os.walk('/content/waste_data'):
    lower_dirs = [d.lower() for d in dirs]
    # Check if this folder contains at least 4 of our 6 known classes
    if sum(wc in lower_dirs for wc in WASTE_CLASSES) >= 4:
        DATA_ROOT = root
        break

if DATA_ROOT is None:
    # fallback: find any folder with 5+ image-containing subfolders
    for root, dirs, files in os.walk('/content/waste_data'):
        valid = [d for d in dirs if any(
            f.lower().endswith(('.jpg','.jpeg','.png'))
            for f in os.listdir(os.path.join(root, d))
        )]
        if len(valid) >= 4:
            DATA_ROOT = root
            break

all_classes = sorted([
    d for d in os.listdir(DATA_ROOT)
    if os.path.isdir(os.path.join(DATA_ROOT, d))
])
NUM_CLASSES  = len(all_classes)
class_to_idx = {c: i for i, c in enumerate(all_classes)}
print(f'✅ Found {NUM_CLASSES} waste categories in: {DATA_ROOT}')
print(f'   Classes: {all_classes}')

# ── 6. Config ─────────────────────────────────────────────────────
IMG_SIZE   = 224
BATCH_SIZE = 32
SEED       = 42
AUTOTUNE   = tf.data.AUTOTUNE

# Eco-friendly colour palette for each waste class (for UI)
CLASS_COLOURS = {
    'cardboard': '#D4875A',   # brown
    'glass'    : '#5BB8D4',   # light blue
    'metal'    : '#8A9BA8',   # steel grey
    'paper'    : '#E8D87A',   # yellow
    'plastic'  : '#E87A7A',   # red
    'trash'    : '#9B9B9B',   # dark grey
}
# Bin emoji for each class
CLASS_EMOJI = {
    'cardboard': '📦', 'glass': '🍾', 'metal': '🥫',
    'paper': '📄', 'plastic': '🧴', 'trash': '🗑️'
}
# Recycling tip for each class
RECYCLE_TIPS = {
    'cardboard': 'Flatten boxes and place in the blue recycling bin. Remove any tape.',
    'glass'    : 'Rinse and place in glass recycling. Do not mix with broken glass.',
    'metal'    : 'Rinse cans and crush to save space. Most metals are 100% recyclable.',
    'paper'    : 'Keep dry and clean. Wet or greasy paper (pizza boxes) goes to compost.',
    'plastic'  : 'Check the recycling number (♳–♷). Rinse before recycling.',
    'trash'    : 'Non-recyclable. Place in general waste bin. Reduce usage where possible.',
}

# ── 7. Collect all image paths & labels ──────────────────────────
IMG_EXTS = ('.jpg', '.jpeg', '.png', '.bmp', '.webp')
all_paths, all_labels = [], []

for cls in tqdm(all_classes, desc='Scanning classes'):
    cls_dir = os.path.join(DATA_ROOT, cls)
    for fname in os.listdir(cls_dir):
        if fname.lower().endswith(IMG_EXTS):
            all_paths.append(os.path.join(cls_dir, fname))
            all_labels.append(class_to_idx[cls])

all_paths  = np.array(all_paths)
all_labels = np.array(all_labels, dtype=np.int32)
print(f'✅ Total images: {len(all_paths)} across {NUM_CLASSES} classes')

# Per-class count
for cls, idx in class_to_idx.items():
    n = int(np.sum(all_labels == idx))
    print(f'   {CLASS_EMOJI.get(cls,"")  } {cls:<12}: {n} images')

# ── 8. Train / Val / Test split (70 / 15 / 15) ───────────────────
X_train, X_tmp, y_train, y_tmp = train_test_split(
    all_paths, all_labels,
    test_size=0.30, random_state=SEED, stratify=all_labels
)
X_val, X_test, y_val, y_test = train_test_split(
    X_tmp, y_tmp,
    test_size=0.50, random_state=SEED, stratify=y_tmp
)
print(f'   Train: {len(X_train)} | Val: {len(X_val)} | Test: {len(X_test)}')

# ── 9. Class weights (dataset is imbalanced — Trash has only 137) ─
cw_vals      = compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
class_weights = dict(enumerate(cw_vals))
print('✅ Class weights computed (handles Trash class imbalance)')

# ── 10. tf.data pipeline with HEAVY augmentation ─────────────────
#    Trash appears in every orientation → aggressive rotation + zoom
def load_image(path, label):
    img = tf.io.read_file(path)
    img = tf.image.decode_image(img, channels=3, expand_animations=False)
    img = tf.image.resize(img, [IMG_SIZE, IMG_SIZE])
    img = tf.cast(img, tf.float32) / 255.0
    return img, label

def augment_heavy(img, label):
    """Heavy augmentation — trash can appear in ANY orientation."""
    # Spatial
    img = tf.image.random_flip_left_right(img)
    img = tf.image.random_flip_up_down(img)
    img = tf.image.rot90(img, k=tf.random.uniform([], 0, 4, dtype=tf.int32))

    # Zoom simulation via random crop + resize
    crop_size = tf.random.uniform([], 180, 224, dtype=tf.int32)
    img = tf.image.random_crop(img, size=[crop_size, crop_size, 3])
    img = tf.image.resize(img, [IMG_SIZE, IMG_SIZE])

    # Colour jitter (lighting & environment variability)
    img = tf.image.random_brightness(img,  0.20)
    img = tf.image.random_contrast(img,    0.80, 1.20)
    img = tf.image.random_saturation(img,  0.75, 1.25)
    img = tf.image.random_hue(img,         0.08)

    # Random Gaussian noise (simulates camera noise)
    noise = tf.random.normal(shape=tf.shape(img), mean=0.0, stddev=0.02)
    img   = img + noise

    return tf.clip_by_value(img, 0.0, 1.0), label

train_ds = (tf.data.Dataset.from_tensor_slices((X_train, y_train))
            .shuffle(2000, seed=SEED)
            .map(load_image,      num_parallel_calls=AUTOTUNE)
            .map(augment_heavy,   num_parallel_calls=AUTOTUNE)
            .batch(BATCH_SIZE)
            .prefetch(AUTOTUNE))

val_ds   = (tf.data.Dataset.from_tensor_slices((X_val, y_val))
            .map(load_image, num_parallel_calls=AUTOTUNE)
            .batch(BATCH_SIZE)
            .prefetch(AUTOTUNE))

test_ds  = (tf.data.Dataset.from_tensor_slices((X_test, y_test))
            .map(load_image, num_parallel_calls=AUTOTUNE)
            .batch(BATCH_SIZE)
            .prefetch(AUTOTUNE))

print('✅ tf.data pipelines ready (heavy augmentation enabled)')

# ── 11. Build EfficientNetB3 model ───────────────────────────────
#    EfficientNetB3 > MobileNetV2 for texture classification (waste materials)
#    B3 hits sweet spot between accuracy and speed for this dataset size
base = EfficientNetB3(
    input_shape=(IMG_SIZE, IMG_SIZE, 3),
    include_top=False,
    weights='imagenet'
)
base.trainable = False

inputs = layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
x      = base(inputs, training=False)
x      = layers.GlobalAveragePooling2D()(x)
x      = layers.BatchNormalization()(x)
x      = layers.Dense(512, activation='relu',
                      kernel_regularizer=regularizers.l2(1e-4))(x)
x      = layers.BatchNormalization()(x)
x      = layers.Dropout(0.45)(x)
x      = layers.Dense(256, activation='relu',
                      kernel_regularizer=regularizers.l2(1e-4))(x)
x      = layers.Dropout(0.30)(x)
outputs = layers.Dense(NUM_CLASSES, activation='softmax')(x)
model   = models.Model(inputs, outputs)

model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-3),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy',
             tf.keras.metrics.SparseTopKCategoricalAccuracy(k=2, name='top2_acc')]
)
print(f'✅ EfficientNetB3 model built — {model.count_params():,} parameters')

# ── 12. Callbacks ─────────────────────────────────────────────────
def make_callbacks(phase):
    return [
        callbacks.EarlyStopping(
            monitor='val_accuracy', patience=6,
            restore_best_weights=True, verbose=1),
        callbacks.ReduceLROnPlateau(
            monitor='val_loss', factor=0.4,
            patience=3, min_lr=1e-7, verbose=1),
        callbacks.ModelCheckpoint(
            f'best_waste_model_phase{phase}.h5',
            monitor='val_accuracy',
            save_best_only=True,
            save_weights_only=False, verbose=1),
        callbacks.CSVLogger(f'waste_log_phase{phase}.csv')
    ]

# ── 13. Phase 1: Train head only (base frozen) ───────────────────
print('\n── Phase 1: Training head (EfficientNetB3 frozen) ──')
history1 = model.fit(
    train_ds, validation_data=val_ds,
    epochs=25, class_weight=class_weights,
    callbacks=make_callbacks(1)
)
print('✅ Phase 1 complete!')

# ── 14. Phase 2: Fine-tune top 40 layers ─────────────────────────
base.trainable = True
for layer in base.layers[:-40]:
    layer.trainable = False

model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-5),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy',
             tf.keras.metrics.SparseTopKCategoricalAccuracy(k=2, name='top2_acc')]
)
print('\n── Phase 2: Fine-tuning top 40 layers of EfficientNetB3 ──')
history2 = model.fit(
    train_ds, validation_data=val_ds,
    epochs=20, class_weight=class_weights,
    callbacks=make_callbacks(2)
)
print('✅ Phase 2 complete!')

# ── 15. Test set evaluation ───────────────────────────────────────
print('\n── Test Set Evaluation ──')
best_model   = tf.keras.models.load_model('best_waste_model_phase2.h5')
test_results = best_model.evaluate(test_ds, verbose=1)
print(f'   Test Accuracy  : {test_results[1]*100:.2f}%')
print(f'   Test Top-2 Acc : {test_results[2]*100:.2f}%')

# Confusion matrix on test set
y_pred_probs = best_model.predict(test_ds, verbose=0)
y_pred       = np.argmax(y_pred_probs, axis=1)
cm           = confusion_matrix(y_test, y_pred)

fig, axes = plt.subplots(1, 3, figsize=(22, 6))

# (a) Confusion matrix
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens',
            xticklabels=all_classes, yticklabels=all_classes,
            ax=axes[0], linewidths=0.5)
axes[0].set_title('Confusion Matrix (Test Set)', fontsize=13)
axes[0].set_xlabel('Predicted'); axes[0].set_ylabel('Actual')

# (b) Accuracy curve
acc     = history1.history['accuracy']     + history2.history['accuracy']
val_acc = history1.history['val_accuracy'] + history2.history['val_accuracy']
ph1_ep  = len(history1.history['accuracy'])
axes[1].plot(acc,     label='Train',      color='steelblue', linewidth=2)
axes[1].plot(val_acc, label='Validation', color='coral',     linewidth=2)
axes[1].axvline(x=ph1_ep, color='gray', linestyle='--', alpha=0.7, label='Fine-tune start')
axes[1].set_title('Model Accuracy', fontsize=13)
axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('Accuracy')
axes[1].legend(); axes[1].grid(True, alpha=0.3)

# (c) Loss curve
loss     = history1.history['loss']     + history2.history['loss']
val_loss = history1.history['val_loss'] + history2.history['val_loss']
axes[2].plot(loss,     label='Train',      color='steelblue', linewidth=2)
axes[2].plot(val_loss, label='Validation', color='coral',     linewidth=2)
axes[2].axvline(x=ph1_ep, color='gray', linestyle='--', alpha=0.7, label='Fine-tune start')
axes[2].set_title('Model Loss', fontsize=13)
axes[2].set_xlabel('Epoch'); axes[2].set_ylabel('Loss')
axes[2].legend(); axes[2].grid(True, alpha=0.3)

plt.suptitle('Waste Segregation AI — EfficientNetB3', fontsize=14, y=1.01)
plt.tight_layout()
plt.savefig('training_curves.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'Best val accuracy : {max(val_acc)*100:.2f}%')
print('\nClassification Report:')
print(classification_report(y_test, y_pred, target_names=all_classes))

# ── 16. Save model & labels ───────────────────────────────────────
best_model.save('waste_classifier_final.h5')
with open('waste_labels.json', 'w') as f:
    json.dump(all_classes, f, indent=2)

print(f'✅ Model saved  : {os.path.getsize("waste_classifier_final.h5")/1e6:.1f} MB')
print(f'✅ Labels saved : {NUM_CLASSES} classes')

# ── 17. Download files ────────────────────────────────────────────
from google.colab import files
files.download('waste_classifier_final.h5')
files.download('waste_labels.json')
files.download('training_curves.png')

print()
print('='*50)
print('  DONE — check your browser downloads!')
print('='*50)
print(f'  Classes   : {NUM_CLASSES}')
print(f'  Val Acc   : {max(val_acc)*100:.2f}%')
print(f'  Test Acc  : {test_results[1]*100:.2f}%')
print(f'  Params    : {model.count_params():,}')
print('='*50)

# ╔══════════════════════════════════════════════════════════════════╗
# ║   PREDICTION WIDGET UI  (runs after training)                   ║
# ╚══════════════════════════════════════════════════════════════════╝

# ── Reload model & labels (safe to re-run) ────────────────────────
model = tf.keras.models.load_model('waste_classifier_final.h5')
with open('waste_labels.json') as f:
    labels = json.load(f)
NUM_CLASSES = len(labels)

# Rebuild lookup dicts using loaded labels
_COLOURS = {
    'cardboard': '#D4875A', 'glass': '#5BB8D4', 'metal': '#8A9BA8',
    'paper': '#E8D87A',     'plastic': '#E87A7A', 'trash': '#9B9B9B'
}
_EMOJI = {
    'cardboard': '📦', 'glass': '🍾', 'metal': '🥫',
    'paper': '📄',    'plastic': '🧴', 'trash': '🗑️'
}
_TIPS = {
    'cardboard': 'Flatten boxes and place in the blue recycling bin. Remove any tape.',
    'glass'    : 'Rinse and place in glass recycling. Do not mix with broken glass.',
    'metal'    : 'Rinse cans and crush to save space. Most metals are 100% recyclable.',
    'paper'    : 'Keep dry and clean. Wet or greasy paper (pizza boxes) goes to compost.',
    'plastic'  : 'Check the recycling number (♳–♷). Rinse before recycling.',
    'trash'    : 'Non-recyclable. Place in general waste bin. Reduce usage where possible.',
}
_DEFAULT_COLOR  = '#718096'
_DEFAULT_EMOJI  = '🗂️'
_DEFAULT_TIP    = 'Dispose of responsibly according to local guidelines.'

print(f'✅ Model reloaded — {NUM_CLASSES} waste classes')

# ── Preprocessing & prediction ────────────────────────────────────
def preprocess_image(img_bytes):
    img = Image.open(io.BytesIO(img_bytes)).convert('RGB')
    img = img.resize((IMG_SIZE, IMG_SIZE), Image.BILINEAR)
    arr = np.array(img, dtype=np.float32) / 255.0
    return np.expand_dims(arr, axis=0)

def predict_waste(img_bytes, top_k=4):
    batch = preprocess_image(img_bytes)
    preds = model.predict(batch, verbose=0)[0]
    top_i = np.argsort(preds)[-top_k:][::-1]
    return [(labels[i], float(preds[i])) for i in top_i]

def pil_to_b64(pil_img, size=(260, 260)):
    pil_img = pil_img.resize(size, Image.BILINEAR)
    buf = io.BytesIO()
    pil_img.save(buf, format='PNG')
    return base64.b64encode(buf.getvalue()).decode()

# ── UI Styles ─────────────────────────────────────────────────────
CARD_STYLE = """
<style>
  @import url('https://fonts.googleapis.com/css2?family=Inter:wght@400;500;600;700&display=swap');
  .waste-wrap  { font-family:'Inter',sans-serif; max-width:600px; }
  .waste-grid  { display:grid; grid-template-columns:260px 1fr; gap:16px;
                 align-items:start; margin-bottom:14px; }
  .img-box     { border-radius:10px; overflow:hidden; border:1px solid #e2e8f0; }
  .img-box img { width:100%; display:block; }
  .img-lbl     { font-size:11px; font-weight:600; color:#718096; text-transform:uppercase;
                 letter-spacing:.06em; padding:6px 10px; background:#f7fafc;
                 border-top:1px solid #e2e8f0; }
  .result-panel { display:flex; flex-direction:column; gap:10px; }
  .top-card    { border-radius:10px; padding:14px 16px; border:1px solid; }
  .top-emoji   { font-size:32px; margin-bottom:4px; }
  .top-class   { font-size:22px; font-weight:700; text-transform:capitalize; }
  .top-conf    { font-size:13px; margin-top:2px; opacity:0.8; }
  .tip-box     { background:#fffbeb; border:1px solid #fbd38d; border-radius:8px;
                 padding:10px 13px; font-size:12px; color:#744210; line-height:1.55; }
  .tip-label   { font-weight:600; margin-bottom:3px; font-size:11px;
                 text-transform:uppercase; letter-spacing:.06em; color:#b7791f; }
  .bars-panel  { background:#f8fafc; border:1px solid #e2e8f0; border-radius:10px;
                 padding:14px 16px; }
  .bars-title  { font-size:11px; font-weight:600; color:#a0aec0; text-transform:uppercase;
                 letter-spacing:.06em; margin-bottom:10px; }
  .bar-row     { display:flex; align-items:center; gap:10px; margin-bottom:8px; }
  .bar-dot     { width:11px; height:11px; border-radius:50%; flex-shrink:0; }
  .bar-name    { font-size:13px; color:#2d3748; width:90px; text-transform:capitalize; }
  .bar-track   { flex:1; background:#e2e8f0; border-radius:6px; height:9px; overflow:hidden; }
  .bar-fill    { height:100%; border-radius:6px; }
  .bar-pct     { font-size:12px; color:#4a5568; width:42px; text-align:right; font-weight:500; }
  .divider     { border:none; border-top:1px solid #e2e8f0; margin:10px 0; }
  .recyclable  { display:inline-block; font-size:11px; font-weight:600; padding:3px 10px;
                 border-radius:20px; margin-top:6px; }
  .yes-recycle { background:#c6f6d5; color:#22543d; }
  .no-recycle  { background:#fed7d7; color:#742a2a; }
</style>
"""

NON_RECYCLABLE = {'trash'}

def recycle_badge(cls_name):
    if cls_name.lower() in NON_RECYCLABLE:
        return '<span class="recyclable no-recycle">✗ Non-Recyclable</span>'
    return '<span class="recyclable yes-recycle">♻ Recyclable</span>'

def result_html(predictions, img_bytes):
    top_name, top_conf = predictions[0]
    key       = top_name.lower()
    color     = _COLOURS.get(key, _DEFAULT_COLOR)
    emoji     = _EMOJI.get(key,   _DEFAULT_EMOJI)
    tip       = _TIPS.get(key,    _DEFAULT_TIP)

    # Lighten colour for card background
    pil_orig  = Image.open(io.BytesIO(img_bytes)).convert('RGB')
    img_b64   = pil_to_b64(pil_orig)

    bars = ""
    for name, conf in predictions:
        pct  = round(conf * 100, 1)
        col  = _COLOURS.get(name.lower(), _DEFAULT_COLOR)
        bars += f"""
        <div class="bar-row">
          <div class="bar-dot" style="background:{col};"></div>
          <span class="bar-name">{name}</span>
          <div class="bar-track">
            <div class="bar-fill" style="width:{pct}%;background:{col};opacity:0.85;"></div>
          </div>
          <span class="bar-pct">{pct}%</span>
        </div>"""

    return f"""
    {CARD_STYLE}
    <div class="waste-wrap">
      <div class="waste-grid">
        <div>
          <div class="img-box">
            <img src="data:image/png;base64,{img_b64}" alt="Uploaded waste image">
            <div class="img-lbl">📷 Uploaded Image</div>
          </div>
        </div>
        <div class="result-panel">
          <div class="top-card"
               style="background:{color}18;border-color:{color}55;color:{color};">
            <div class="top-emoji">{emoji}</div>
            <div class="top-class">{top_name}</div>
            <div class="top-conf">Confidence: <b>{round(top_conf*100, 2)}%</b></div>
            {recycle_badge(top_name)}
          </div>
          <div class="tip-box">
            <div class="tip-label">♻ Disposal Tip</div>
            {tip}
          </div>
        </div>
      </div>
      <div class="bars-panel">
        <div class="bars-title">All class probabilities</div>
        {bars}
      </div>
    </div>"""

# ── Build Widget UI ───────────────────────────────────────────────
title_html = HTML("""
<div style="font-family:'Inter',sans-serif; max-width:600px; margin-bottom:8px;">
  <h2 style="margin:0 0 4px; font-size:22px; color:#1a202c; font-weight:700;">
    ♻ Waste Segregation AI — Eco-Sorter
  </h2>
  <p style="margin:0; font-size:14px; color:#718096;">
    Upload a photo of any waste item — the model classifies it into
    <b>Cardboard, Glass, Metal, Paper, Plastic</b> or <b>Trash</b>
    and tells you exactly how to dispose of it.
  </p>
</div>
""")

uploader  = FileUpload(
    accept='.jpg,.jpeg,.png,.webp,.bmp',
    multiple=False,
    description='Choose waste image',
    layout=Layout(width='220px')
)
run_btn   = Button(
    description='Classify Waste',
    button_style='success',
    icon='recycle',
    layout=Layout(width='155px', height='36px')
)
clear_btn = Button(
    description='Clear',
    button_style='',
    icon='times',
    layout=Layout(width='90px', height='36px')
)
status = HTML("")
out    = Output()

def on_classify(b):
    out.clear_output()
    status.value = ""
    if not uploader.value:
        status.value = '<span style="color:#c53030;font-size:13px;">⚠ No image uploaded.</span>'
        return
    file_info = list(uploader.value.values())[0]
    img_bytes = bytes(file_info['content'])
    fname     = file_info['metadata']['name']
    status.value = (f'<span style="color:#3182ce;font-size:13px;">'
                    f'Classifying <b>{fname}</b>…</span>')
    run_btn.disabled = True
    try:
        predictions = predict_waste(img_bytes)
        with out:
            ipd.display(HTML(result_html(predictions, img_bytes)))
        top_name, top_conf = predictions[0]
        status.value = (f'<span style="color:#276749;font-size:13px;">'
                        f'✅ Done — detected <b>{top_name}</b> '
                        f'({round(top_conf*100,1)}% confidence)</span>')
    except Exception as e:
        status.value = f'<span style="color:#c53030;font-size:13px;">Error: {e}</span>'
        import traceback
        with out:
            print(traceback.format_exc())
    finally:
        run_btn.disabled = False

def on_clear(b):
    out.clear_output()
    status.value = ""

run_btn.on_click(on_classify)
clear_btn.on_click(on_clear)

ui = VBox([
    title_html,
    HBox([uploader, run_btn, clear_btn],
         layout=Layout(gap='10px', align_items='center')),
    status,
    out
], layout=Layout(gap='10px', padding='16px 0'))

ipd.display(ui)
```
